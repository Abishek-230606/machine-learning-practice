# SINGLE LAYER FEED FORWARD NETWORK

# Scaling the Matrix Dimensions

### In a multiple-input neuron model, your inputs R = 3 and your number of neurons S = 5.

### Input Vector ($\mathbf{p}$): Remains a $3 \times 1$ column vector.

### Weight Matrix ($\mathbf{W}$): Since you have $5$ neurons and each needs $3$ weights, your weight matrix must be of size $S \times R$, which is $5 \times 3$.

### Bias Vector ($\mathbf{b}$): Since you have $5$ neurons, you need $5$ biases. This becomes an $S \times 1$ column vector, or $5 \times 1$

### When you take the dot product of $\mathbf{W}$ ($5 \times 3$) and $\mathbf{p}$ ($3 \times 1$), the inner dimensions ($3$) collapse, leaving you with a $5 \times 1$ vector for your net input $\mathbf{n}$. You then add your $5 \times 1$ bias vector $\mathbf{b}$.

## Mathematically, the equation $\mathbf{a} = f(\mathbf{Wp} + \mathbf{b})$ computes the outputs for all $5$ neurons simultaneously


```python
def logsig(n):
    return 1/(1+ np.exp(-n))

# R=3 (inputs) , S=5 (nurons)

p = np.array([[0.5],
              [0.8],
              [0.2]])

np.random.seed(42)

# W is now a 5*3 matrix so,
W = np.random.rand(5,3)


# b is now 5*1 coloumn vector

b = np.random.rand(5,1)

# forward pass of the nuron

n = np.dot(W,p)+b
a = logsig(n)

print("Output vector a or the output of the nuron is : \n")
print(n)
print("\n Output vector a or the output of the nuron after activation : \n")
print(a)
print("\nShape of a:", a.shape)
```

They are meaningless because my weight matrix $\mathbf{W}$ and bias vector $\mathbf{b}$ are completely random. The network has no experience and has not been trained. To make these outputs meaningful, you must introduce a target output, calculate the error between your network's prediction and the actual truth, and then adjust those weights and biases using a training algorithm. That iterative weight adjustment is what we call "learning".

# MULTI LAYERED FEED FORWARD NETWORK

To make a network "multi-layer," you simply take the output vector of your first layer and feed it directly as the input vector to the next layer. The mathematical representation for a two-layer network is $\mathbf{a}^2 = f^2(\mathbf{W}^2\mathbf{a}^1 + \mathbf{b}^2)$, where $\mathbf{a}^1$ is the output from the first layer.

```python

# Assume 'a' from our previous code is the output of Layer 1 (Hidden Layer)
a1 = a # Shape: (5, 1)

# Layer 2 (Output Layer): Let's map those 5 hidden neurons to 2 final output neurons
W2 = np.random.rand(2, 5) # 2 neurons, receiving 5 inputs each
b2 = np.random.rand(2, 1) # 2 biases for 2 neurons

# Forward pass for Layer 2
n2 = np.dot(W2, a1) + b2
a2 = logsig(n2)

print("Multi-Layer Final Output a2:\n", a2)
```
