Key Idea

Pandas was created because Python's built-in data structures become inefficient and cumbersome for large-scale tabular data analysis.

Why It Exists

To provide fast, readable, and powerful tools for manipulating structured data.

Real-World Use

Cleaning, analyzing, transforming, and preparing data before visualization or machine learning.

Common Mistake

Thinking Pandas is "just for reading CSV files." Reading data is only the first step; its real strength is transforming and analyzing data efficiently.

Engineer Handbook – Lesson 2
Key Idea

A DataFrame represents a complete labeled table, while a Series represents a single labeled column (or one-dimensional data).

Why They Exist

To model structured data in a way that's readable, efficient, and easy to manipulate.

Real-World Use

Customer records, sales transactions, medical reports, financial data, IoT sensor readings—almost any structured dataset can be represented as a DataFrame.

Common Mistake

Thinking a DataFrame is "just a list of lists." It also stores labels (indexes and column names), metadata, and uses optimized structures under the hood.

```python
import pandas as pd

students = {
    "Name": ["Arun", "Priya", "Rahul", "Sneha", "Vikram"],
    "Age": [20, 21, 19, 22, 20],
    "Department": ["CSE", "ECE", "AI", "CSE", "IT"],
    "CGPA": [8.5, 9.1, 8.8, 8.2, 7.9]
}

df = pd.DataFrame(students)

name = df["Name"]
print(name)
cgpa = df["CGPA"]
print(cgpa)
name_dep = df[["Name","Department"]]
print(name_dep)
age_cgpa = df[["Age","CGPA"]]
print(age_cgpa)
```

```python
import pandas as pd


students = {
    "Name": ["Arun", "Priya", "Rahul", "Sneha", "Vikram"],
    "Age": [20, 21, 19, 22, 20],
    "Department": ["CSE", "ECE", "AI", "CSE", "IT"],
    "CGPA": [8.5, 9.1, 8.8, 8.2, 7.9]
}

df = pd.DataFrame(students)


res1 = df[df["Department"].isin(["CSE", "AI","IT"])]
print(res1)

res2 = df[df["Age"].between(20,22)]
print(res2)

res3 = df[df["Name"].str.contains("a" , case=False)]
print(res3)

res4 = df[df["Name"].str.startswith("A")]
print(res4)

res5 = df[df["Name"].str.endswith("i")]
print(res5)

res6 = df.query("Age >= 20 and CGPA > 8")
print(res6)
```
